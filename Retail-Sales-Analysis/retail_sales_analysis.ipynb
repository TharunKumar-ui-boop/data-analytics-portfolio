```python
#  Data Load 

import pandas as pd

df = pd.read_excel("Project_01_Saler_Data.xlsx")
```

```python
df
```

```python
# Duplicate remove

df = df.drop_duplicates()
```

```python
df
```

```python
# Fix date colum

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors ='coerce')
```

```python
df = df.dropna(subset=['Order_Date'])
```

```python
df
```

```python
df['Profit'] = df['Sales']-df['Cost']
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
```

```python
df.to_csv('Project_01_Saler_Cleaned_Data.csv',index = False)
```
